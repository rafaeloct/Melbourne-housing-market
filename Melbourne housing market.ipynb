```python
less = missing_values[missing_values < 17000].index
over = missing_values[missing_values >= 17000].index
```

```python
#Contoh mengisi nilai yang hilang dengan median untuk kolom numerik
numeric_features = train[less].select_dtypes(include=['number']).columns
train[numeric_features] = train[numeric_features].fillna(train[numeric_features].median())
```

```python
kategorical_features = train[less].select_dtypes(include=['object']).columns

for column in kategorical_features:
    train[column] = train[column].fillna(train[column].mode()[0])
```

```python
#Menghapus kolom dengan terlalu banyak nilai yang hilang
df = train.drop(columns=over)
```

```python
#Contoh sederhana untuk mengidentifikasi outliers menggunakan IQR
Q1 = df[numeric_features].quantile(0.25)
Q3 = df[numeric_features].quantile(0.75)
IQR = Q3 - Q1
```

```python
# Filter dataframe untuk hanya menyimpan baris yang tidak mengandung outliers pada kolom numerik
condition = ~((df[numeric_features] < (Q1 - 1.5 * IQR)) | (df[numeric_features] > (Q3 + 1.5 * IQR))).any(axis=1)
df_filtered_numeric = df.loc[condition, numeric_features]

# Menggabungkan kembali dengan kolom kategorikal
categorical_features = df.select_dtypes(include=['object']).columns
df = pd.concat([df_filtered_numeric, df.loc[condition, categorical_features]], axis=1)
```

```python
median = df['Price'].median()
df['Price'] = df['Price'].apply(lambda x: median if x < (Q1['Price'] - 1.5 * IQR['Price']) or x > (Q3['Price'] + 1.5 * IQR['Price']) else x)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

# 1. Pisahkan kolom numerik dan kategorikal
numeric_features = X.select_dtypes(
    include=['int64', 'float64']
).columns

categorical_features = X.select_dtypes(
    include=['object', 'category']
).columns

# 2. Preprocessing kolom numerik
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# 3. Preprocessing kolom kategorikal
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(
        strategy='constant',
        fill_value='missing'
    )),
    ('onehot', OneHotEncoder(
        handle_unknown='ignore',
        sparse_output=False
    ))
])

# 4. Gabungkan preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)

print("Preprocessing berhasil dibuat!")
```

```python
# MODEL 1 - LINEAR REGRESSION
linear_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LinearRegression())
])

# Training
linear_model.fit(X_train, y_train)

# Prediksi
y_pred_linear = linear_model.predict(X_test)
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
# MODEL 2 - RANDOM FOREST REGRESSOR
random_forest_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1))
])

# Training
random_forest_model.fit(X_train, y_train)

# Prediksi
y_pred_rf = random_forest_model.predict(X_test)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Membersihkan target yang kosong
mask = y_train.notna()
X_train_clean = X_train.loc[mask]
y_train_clean = y_train.loc[mask]

# Memisahkan kolom teks dan angka
categorical_cols = X_train_clean.select_dtypes(
    include=["object", "category", "string"]
).columns.tolist()

numerical_cols = X_train_clean.select_dtypes(
    include=["number"]
).columns.tolist()

# Preprocessing
preprocessor = ColumnTransformer([
    ("cat", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ), categorical_cols),
    ("num", "passthrough", numerical_cols)
])

# MODEL 3 - HISTOGRAM GRADIENT BOOSTING REGRESSOR
hist_gradient_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", HistGradientBoostingRegressor(
        max_iter=100,
        random_state=42
    ))
])

# Training
hist_gradient_model.fit(X_train_clean, y_train_clean)

# Prediksi
y_pred_hgb = hist_gradient_model.predict(X_test)

print("Training MODEL 3 berhasil!")
print("Jumlah prediksi:", len(y_pred_hgb))
```

```python
import pandas as pd
import numpy as np

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import HistGradientBoostingRegressor

# Menentukan kolom numerik dan kategori
numeric_cols = X_train.select_dtypes(include=["number"]).columns
categorical_cols = X_train.select_dtypes(
    include=["object", "category", "string"]
).columns

# Preprocessing data
preprocessor = ColumnTransformer([
    ("num", SimpleImputer(strategy="median"), numeric_cols),
    ("cat", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        ))
    ]), categorical_cols)
])

# MODEL 3 - HISTOGRAM GRADIENT BOOSTING
hist_gradient_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", HistGradientBoostingRegressor(
        max_iter=20,
        random_state=42,
        early_stopping=True
    ))
])

# Menghapus target kosong jika ada
mask = y_train.notna()
X_train_clean = X_train.loc[mask]
y_train_clean = y_train.loc[mask]

# Training dan prediksi
hist_gradient_model.fit(X_train_clean, y_train_clean)
y_pred_hgb = hist_gradient_model.predict(X_test)

print("Training berhasil!")
print("Jumlah prediksi:", len(y_pred_hgb))
print("Contoh prediksi:", y_pred_hgb[:5])
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import HistGradientBoostingRegressor

# Memisahkan kolom teks dan angka
categorical_cols = X_train.select_dtypes(
    include=["object", "category", "string"]
).columns.tolist()

numerical_cols = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

# Mengubah data teks menjadi angka
preprocessor = ColumnTransformer([
    ("cat", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ), categorical_cols),
    ("num", "passthrough", numerical_cols)
])

# Membuat model
hist_gradient_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", HistGradientBoostingRegressor(
        max_iter=20,
        random_state=42,
        early_stopping=True
    ))
])

# Training dan prediksi
hist_gradient_model.fit(X_train, y_train)
y_pred_hgb = hist_gradient_model.predict(X_test)

print("Training berhasil!")
print("Jumlah prediksi:", len(y_pred_hgb))
print("Contoh prediksi:", y_pred_hgb[:5])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# FUNGSI EVALUASI MODEL REGRESI
def evaluasi_model(nama_model, y_test, y_pred):
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)
    
    print("\n" + "=" * 60)
    print(nama_model)
    print("=" * 60)
    
    print(f"MAE  : {mae:.4f}")
    print(f"MSE  : {mse:.4f}")
    print(f"RMSE : {rmse:.4f}")
    print(f"R²   : {r2:.4f}")
    
    return [mae, mse, rmse, r2]
```

```python
# Evaluasi Model 1: Linear Regression
evaluasi_model("Linear Regression", y_test, y_pred_linear)

# Evaluasi Model 2: Random Forest Regressor
evaluasi_model("Random Forest Regressor", y_test, y_pred_rf)

# Evaluasi Model 3: Hist Gradient Boosting Regressor
evaluasi_model("Hist Gradient Boosting Regressor", y_test, y_pred_hgb)
```

```python
# PERBANDINGAN SEMUA MODEL REGRESI
hasil_perbandingan = pd.DataFrame(
    [
        evaluasi_model("Linear Regression", y_test, y_pred_linear),
        evaluasi_model("Random Forest Regressor", y_test, y_pred_rf),
        evaluasi_model("Hist Gradient Boosting Regressor", y_test, y_pred_hgb)
    ],
    columns=[
        "MAE",
        "MSE",
        "RMSE",
        "R²"
    ],
    index=[
        "Linear Regression",
        "Random Forest Regressor",
        "Hist Gradient Boosting Regressor"
    ]
)

print("\n" + "=" * 60)
print("PERBANDINGAN PERFORMA MODEL REGRESI")
print("=" * 60)

display(hasil_perbandingan.round(4))
```
